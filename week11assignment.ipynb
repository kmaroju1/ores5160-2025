Q2. Basic Statistical Analysis (pandas)

	1.	Compute the average population across all states.

	2.	Find which region has the highest total population.

	3.	Plot a bar chart showing the population by region (use matplotlib)

```python
import pandas as pd
```

```python
filename = "est2018-pop-res (1).csv"
df = pd.read_csv(filename, encoding='latin-1')
```

```python
population_column = 'POPESTIMATE2018'
```

```python
average_population = df[population_column].mean()
print(f"Average population across all states: {average_population:,.0f}")
```

```python
region_column = 'REGION'
```

```python
region_population = df.groupby(region_column)[population_column].sum()
```

```python
highest_region = region_population.idxmax()
highest_population = region_population.max()

print("\nRegion with the highest total population:")
print(f"{highest_region} — {highest_population:,.0f}")
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8, 5))
region_population.sort_values(ascending=False).plot(kind='bar', color='skyblue')
plt.title('Total Population by Region (2018)')
plt.xlabel('Region')
plt.ylabel('Total Population')
plt.tight_layout()
plt.show()
```

Q3. Parsing the JSON File (FHIR structure)

Using json module:

	1.	Load patient.json into a Python dictionary.

	2.	Extract and print:

	•	Patient’s full official name

	•	Gender and birth date

	•	Current address (city, state, postal code)
  
	3.	Count how many phone numbers the patient has, and list their types (e.g., work, mobile, home).

```python
from google.colab import files

uploaded = files.upload()
```

```python
import json
```

```python
filename = "patient (1).json"

with open(filename, "r", encoding="utf-8") as file:
    patient_data = json.load(file)
```

```python
print(type(patient_data))
print("Top-level keys:", list(patient_data.keys()))
```

```python
official_name = patient_data["name"][0]
full_name = " ".join(official_name["given"]) + " " + official_name["family"]
gender = patient_data.get("gender", "Not specified")
birth_date = patient_data.get("birthDate", "Not specified")
address = patient_data["address"][0]
city = address.get("city", "Unknown")
state = address.get("state", "Unknown")
postal_code = address.get("postalCode", "Unknown")
```

```python
print("Patient’s Full Official Name:", full_name)
print("Gender:", gender)
print("Birth Date:", birth_date)
print(f"Address: {city}, {state} {postal_code}")
```

```python
telecom_entries = patient_data.get("telecom", [])
phone_numbers = [t for t in telecom_entries if t.get("system") == "phone"]
num_phones = len(phone_numbers)
phone_types = [t.get("use", "unspecified") for t in phone_numbers]
print(f"Total phone numbers: {num_phones}")
print("Phone types:", phone_types)
```
